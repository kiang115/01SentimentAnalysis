```python
import pandas as pd

# 定义文件路径（替换为你的实际路径）
input_csv = "waimai.csv"
output_csv = "waimai_1k.csv"

# 读取CSV文件
df = pd.read_csv(input_csv)

# 按label=0抽取500行，label=1抽取500行
df_label_0 = df[df['label'] == 0].sample(n=500)
df_label_1 = df[df['label'] == 1].sample(n=500)

# 合并抽取的两行数据
sampled_df = pd.concat([df_label_0, df_label_1])

# 保存为新的CSV文件
sampled_df.to_csv(output_csv, index=False)
```
