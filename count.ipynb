```python
import pandas as pd

# 读取CSV文件
df = pd.read_csv('online-shopping.csv')  # 替换为你的文件路径

# 统计中文字符数超过128的评论比例
def count_chinese(text):
    if pd.isna(text):
        return 0
    return sum(1 for char in str(text) if '\u4e00' <= char <= '\u9fff')

# 应用函数统计每行的中文字符数
df['chinese_count'] = df['review'].apply(count_chinese)

# 计算超过128字符的行数
over_128 = (df['chinese_count'] > 256).sum()
total = len(df)

# 计算比例
ratio = over_128 / total if total > 0 else 0

print(f"总评论数: {total}")
print(f"中文字符超过256的评论数: {over_128}")
print(f"比例: {ratio:.2%}")
```
